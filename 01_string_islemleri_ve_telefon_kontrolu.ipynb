```python
# Kullanıcıdan veriyi al ve değişkene ata
kullanici_posta = input("e-postanızı giriniz: ")

# 'in' kullanarak @ işareti var mı kontrol et
print("@" in kullanici_posta)

#  Python'da bir karakterin kaçıncı sırada olduğunu bulmak için
#    .index() metodu kullanılır.
#  Python'da dilimleme yaparken, eğer metnin en sonuna kadar gitmek istiyorsan,
#  bitiş indeksini hesaplamana hiç gerek yoktur.
#  İkinci kısmı boş bırakırsan Python otomatik olarak en sona kadar gider.

konum_baslangıcı = kullanici_posta.index("@") +1
print(kullanici_posta[konum_baslangıcı:])
print(kullanici_posta.lower())



```

```python
telefon_nu = input("Telefon numaranızı giriniz: ")

# Numarayı temizliyoruz
temizlenmis_nu = telefon_nu.replace(" ", "").replace("-", "")
print("Temizlenmiş numara:", temizlenmis_nu)

# Temizlenmiş numaradan ülke kodunu (90) çekiyoruz
print("Ülke telefon kodu : " , temizlenmis_nu[1:3])

# Orijinal numarada + işareti var mı kontrol ediyoruz
print("+" in telefon_nu)

```
