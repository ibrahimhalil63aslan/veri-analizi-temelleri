***LİST dikkat edilmesi gereken noktaları***

Listelerin dinamik yapısı, onları esnek ve kullanışlı kılar. Ancak, bu esneklik performans maliyetleri de getirir. Özellikle büyük veri setleriyle çalışırken, liste büyütme işlemleri arka planda taşınma ve kopyalama gerektirdiği için pahalı olabilir. Bu nedenle, performansın kritik olduğu durumlarda, sabit tipte veri yapıları (örneğin NumPy dizileri) tercih edilebilir.

Python listelerinin bir diğer önemli özelliği, heterojen yapıda olmalarıdır. Yani, bir liste içinde farklı türde veriler (sayılar, metinler, objeler vb.) saklanabilir. Bu esneklik, listeleri çok yönlü kılar, ancak her elemanın tipini ayrı ayrı yönetmek gerektiğinden, bu durum hızdan ödün vermemize neden olabilir.

Listeler, elemanları referans olarak sakladıkları için, referans tuzağı adı verilen bir duruma da yol açabilir. Bu, listede saklanan bir objenin başka bir yerde değiştirildiğinde, listenin de bu değişiklikten etkilenmesi anlamına gelir. Bu nedenle, listelerle çalışırken bu davranışın farkında olmak önemlidir.

```python
# İçinde aynı ürünlerin tekrar ettiği bir liste:
urunler_listesi = ["Laptop", "Fare", "Laptop", "Klavye", "Fare"]

# Liste -> Set dönüşümü yapıyoruz:
benzersiz_urunler = set(urunler_listesi)

print(benzersiz_urunler)
# Çıktı: {'Laptop', 'Fare', 'Klavye'} (Tekrarlar otomatik silindi!)
```

**List Comprehension**

Döngüleri Tek Satıra İndirger: Normalde 3-4 satır sürecek for döngüsü ve .append() işlemlerini tek satırda tertemiz yazmanı sağlar.

Hızlıdır (Performans): Arka planda C seviyesinde optimize edildiği için geleneksel Python for döngülerine göre daha hızlı çalışır.

Filtreleme ve Dönüştürme Yapabilir: Liste üzerindeki her elemana matematiksel bir işlem uygulayabilir VEYA elemanları belli bir if koşuluna göre süzebilir.

Elemanları Dönüştürme (**Mapping**)

```python
prices= [10,20,30,40]
double_prices = [price * 2 for price in prices ]
print(double_prices)
```

```python
tasks= ["meil","toplantı"]

tasks.append("sunum")
print(tasks)

```

```python
done_task = tasks.pop()
print(done_task)
print(tasks)
```

Elemanları Filtreleme (**Filtering**)

```python
numbers = [1,2,3,4,5,6]
# if x % 2==0 = x iki bölünce kalan 0 olsun,demektir.Yani çift sayıları al,demek.
evens = [x for x in numbers if x % 2 == 0]
print(evens)
```

```python
# Her elemanın karesini alma
sayilar = [1, 2, 3, 4, 5]

# List Comprehension ile:
kareler = [x ** 2 for x in sayilar]

print(kareler)
```

Elektrik Tüketim Verilerini Filtreleme (Filtering)

```python
tuketimler = [1200, 850, 2100, 450, 3200, 1900, 950, 2800]
ilk_yarisi = tuketimler[0:5]
print(ilk_yarisi)
```

```python
# Orijinal tüketim verileri listesi
tuketimler = [1200, 850, 2100, 450, 3200, 1900, 950, 2800]

# List Comprehension ile süzme işlemi
yuksek_tuketim = [tuketim for tuketim in tuketimler if tuketim >= 1500]

print(yuksek_tuketim)
```

Elektrik Faturasına Zam/KDV Uygulama (Mapping)

```python
# Meskenin son 6 aylık elektrik faturası tutarları (TL cinsinden)

faturalar = [350.0, 420.5, 290.0, 510.0, 380.0, 600.0]

# faturalar listesindeki her bir 'fatura'yı al ve 1.15 ile çarp
zamli_faturalar = [round(fatura * 1.15, 2) for fatura in faturalar]


print(zamli_faturalar)
```
