```python
markalar = {"VW", "Audi", "Bentley", "Aston Martin","BMW","BMW"}
print(markalar)
```

```python
# Eleman ekleme--add
markalar.add("mercedes")
print(markalar)
```

```python
# Birden fazla eklemeler için---update([])
markalar.update(["TOGG","Ford"])
print(markalar)
```

```python
# Silmek için---remove()
markalar.remove("VW")
print(markalar)
```

```python
# Silmek istenen setin içinde yoksa?---Setler hata vermiyor.
markalar.discard("mavi")
print(markalar )
```

```python
# Setlerde sıra numarası olmadığı için silmek için pop() aşağıdaki gibi kullanılır.
Silinen =markalar.pop()
print("Silinen :" , Silinen )
#Setteki RASTGELE  eleman silinir.
```

```python
# Kesişim-----intersection
avrupa = {"İtalya","Fransa","İspanya"}
balkan = {"Türkiye","Yunanistan","İtalya"}
print(avrupa.intersection(balkan))
```

```python
# Fark-----difference
print(avrupa.difference(balkan))
```

```python
## Birleştirme---union
##  iki setteki benzersiz elemanlar birleştirilir.
print(avrupa.union(balkan))
```
