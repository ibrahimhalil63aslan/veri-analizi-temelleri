```python
# ==============================================================================
# DEDAŞ Tarımsal Sulama Tüketim ve Kaçak/Anormallik Analiz Simülasyonu
# ==============================================================================

# Abone ve Sayaç Parametreleri
abone_no = "TS-84920"
sozlesme_gucu_kw = 50.0  # kW cinsinden sözleşme gücü
onceki_endeks = 14500.0  # kWh cinsinden geçen ayki okuma
guncel_endeks = 18200.0  # kWh cinsinden bu ayki okuma
cks_arazi_donum = 100    # Çiftçi Kayıt Sistemi'ndeki arazi büyüklüğü (Dönüm)
sezon_aktif_mi = True    # Yaz/Sulama dönemi mi?

# 1. Aylık Tüketim Hesabı
aylik_tuketim_kwh = guncel_endeks - onceki_endeks

# 2. Endeks Hata Kontrolü
if guncel_endeks < onceki_endeks:
    print(f"[HATA] Abone {abone_no}: Güncel endeks öncekinden küçük! Sayaç ters dönmüş veya hatalı okuma yapılmış.")
else:
    # 3. Teorik Maksimum Tüketim Limiti (Güç x 24 saat x 30 gün)
    max_teorik_tuketim = sozlesme_gucu_kw * 24 * 30

    # Anormallik ve Kaçak Şüphe Bayrakları
    is_emri_gerekli = False
    tespit_nedeni = ""

    # 4. Analiz ve Koşul Kontrolleri
    if aylik_tuketim_kwh > max_teorik_tuketim:
        is_emri_gerekli = True
        tespit_nedeni = "Sözleşme gücü üzerinde aşırı tüketim (Kapasite aşımı / Trafosuz kaçak şüphesi)"

    elif sezon_aktif_mi and (aylik_tuketim_kwh < (cks_arazi_donum * 10)):
        # Sulama sezonunda 100 dönüm araziye göre tüketim mantıksız derecede düşükse
        is_emri_gerekli = True
        tespit_nedeni = "ÇKS arazisine göre sulama sezonunda şüpheli düşük tüketim (Harici hat / Müdahale şüphesi)"

    # 5. Raporlama Çıktısı
    print("==================================================")
    print("      DEDAŞ TARIMSAL SULAMA ANALİZ RAPORU        ")
    print("==================================================")
    print(f"Abone Numarası    : {abone_no}")
    print(f"ÇKS Arazi Büyüklüğü: {cks_arazi_donum} Dönüm")
    print(f"Sözleşme Gücü    : {sozlesme_gucu_kw} kW")
    print(f"Aylık Tüketim    : {aylik_tuketim_kwh:.1f} kWh")
    print("--------------------------------------------------")

    if is_emri_gerekli:
        print("[DURUM]    : ⚠️ ŞÜPHELİ / ANORMALLİK TESPİT EDİLDİ")
        print(f"[AÇIKLAMA] : {tespit_nedeni}")
        print("[AKSİYON]  : Otomatik Saha İnceleme İş Emri Oluşturuldu.")
    else:
        print("[DURUM]    :   NORMAL")
        print("[AKSİYON]  : Faturalandırma onaylandı.")
    print("==================================================")
```
