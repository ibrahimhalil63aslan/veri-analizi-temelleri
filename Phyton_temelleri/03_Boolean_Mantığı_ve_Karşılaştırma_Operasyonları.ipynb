
**short-circuit**

and / or Operatörleri
Bu operatörler cevabı True veya False yapmakla uğraşmaz. Sadece değerlerin üzerinden atlayarak son durdukları değeri kopyalayıp sana verirler.


and       Kuralı: "İlk Yanlışı Bul ya da Sonuna Kadar Git"
and       gördüğünde Python soldan sağa bakar.

İlk gördüğü değer Boş (False) ise hemen durur ve o boş değeri fırlatır.

İlk değer Dolu (True) ise "Tamam bu geçti" der ve ikinci değeri fırlatır.

```python
'' and 'y'   # İlk değer BOŞ (''). Python direkt '' döndürür. (Çıktı: '')
```

```python
' ' and 'y'  # İlk değer DOLU (' '). Python ikinciye geçer ve 'y' döndürür. (Çıktı: 'y')
```

**or Kuralı**

"İlk Doğruyu Bul ya da Sonuna Kadar Git"
or gördüğünde Python ilk Dolu (True) değeri arar.

İlk gördüğü değer Dolu (True) ise "Harika, buldum!" der ve o dolu değeri fırlatır.

İlk değer Boş (False) ise mecbur ikinciye geçer ve ikinci değeri fırlatır.

```python
'' or 'y'    # İlk değer BOŞ (''). Python ikinciye geçer ve 'y' döndürür. (Çıktı: 'y')
```

```python
'a' or 'y'   # İlk değer DOLU ('a'). Python direkt 'a' döndürür. (Çıktı: 'a')
```
