```python
# Tuple içinde farklı türler bulundurlabilir.
kisi =("İbrahim Halil ASLAN",26,True)
print(kisi)
```

```python
# Tuple değiişmez  ama içindeki listelerdeğişebilir.
bilgi = ("Türkiye",["İstanbul","Ankara"])
bilgi[1].append("İzmir")
print(bilgi)
```

```python
# Parantezsiz tuplede tanımlanabilir
konum=40.1,29.0
print(type(konum))
```

```python
# Unpacking işlemi ile birden fazla değeri bir seferde değişkenlere atamak için kullanılabilir
enlem,boylam =konum
print("Enlem:", enlem,"Boylam:",boylam)
```

```python
# Birleştirme
a=(1,2)
b=(3,4)
print(a + b) #---Yeni Tuple oluşturuldu ama eksi Tupleler değişmedi.
```

```python
harfler=("a","b","c","d","a","b","c")
print(harfler.count("a"))# ---- kaç tane a var olduğunu gösterir.
print(harfler.index("a"))# ---- kaçıncı indexte a var olduğunu gösterir.
```

```python
# Table, Immutable (Değişmez) bir yapı ama çoğaltılabilir.
print(("Phyton Coderspace",)*3)
```

```python
# 1. Tuple oluşturma
order = (1234, 5678, 2450.0)

# 2. Unpacking
order_id, customer_id, total = order

# 3.  sınıflandırma yapıldı.
durumlar = ["Düşük", "Yüksek"]
print(durumlar[total > 1000])
```
