```python
prices = {
    "laptop": 25000,
    "telefon": 15000,
    "kulaklik": 1500,
}

# Dictionary comprehension ile yeni sözlüğü oluşturma
indirimli_fiyatlar = {urun: fiyat * 0.9 for urun, fiyat in prices.items()}

# Her ürün için istenen formatta çıktı alma
for urun, eski_fiyat in prices.items():
    yeni_fiyat = indirimli_fiyatlar[urun]
    print(f"{urun}: {eski_fiyat} -> {yeni_fiyat}")
```

```python
# 1. ORİJİNAL VERİ
# Elektrik abonelerinin net fatura tutarlarını tutan varsayılan sözlük
faturalar = {
    "Abone_A": 1000,
    "Abone_B": 2500,
    "Abone_C": 800
}

# 2. DICTIONARY COMPREHENSION (TEK SATIRDA SÖZLÜK OLUŞTURMA)
# .items() ile faturalar sözlüğündeki her bir (abone, fiyat) çiftini alıyoruz.
# Fiyatı 1.18 ile çarparak %18 KDV ekliyoruz ve 'kdvli_faturalar' adında yeni bir sözlük oluşturuyoruz.
kdvli_faturalar = {abone: fiyat * 1.18 for abone, fiyat in faturalar.items()}

# 3. VERİLERİ KARŞILAŞTIRMALI YAZDIRMA
# Orijinal 'faturalar' sözlüğündeki her aboneyi ve eski fiyatı döngüye alıyoruz.
for abone, eski in faturalar.items():
    # 'kdvli_faturalar' sözlüğünden o aboneye ait hesaplanmış yeni fiyatı çekiyoruz.
    yeni = kdvli_faturalar[abone]

    # f-string formatı ile ekran çıktısını veriyoruz.
    print(f"{abone}: {eski} -> {yeni}")

```
