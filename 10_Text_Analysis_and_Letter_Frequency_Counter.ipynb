```python
# Müşterilerden gelen kelime logları
kelimeler = ["harika", "hizmet", "yavaş", "harika", "kötü", "hizmet", "harika"]

# Frekansları tutacağımız boş defter (sözlük)
kelime_sayaci = {}

for kelime in kelimeler:
    # Sözlükte yoksa 0 al, varsa mevcut sayısını al, üstüne 1 ekle
    kelime_sayaci[kelime] = kelime_sayaci.get(kelime, 0) + 1

print("Kelime Frekans Tablosu:")
print(kelime_sayaci)
```

```python
cumle = input().lower()
harf_frekansi = {}

for harf in cumle:
    # 1. Kural: Boşlukları atla
    if harf == " ":
        continue

    # 2. Kural: Frekansı say / güncelle
    harf_frekansi[harf] = harf_frekansi.get(harf, 0) + 1

# 3. Kural: Büyükten küçüğe sırala
sirali_harfler = sorted(harf_frekansi.items(), key=lambda x: x[1], reverse=True)

# 4. Kural: Sadece ilk 3 harfi ve frekansını saf bir şekilde yazdır (Başlık yok!)
for harf, sayi in sirali_harfler[:3]:
    print(f"{harf}: {sayi}")
```
