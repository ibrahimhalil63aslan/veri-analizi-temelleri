```python
# Global function---> Bunun için fonksiyon global taanımlanmalı yani fonksiyonun dışında tanımlanmalıdır.
seker_sayısı = 89

def seker_ekle():
  global seker_sayısı
  seker_sayısı = seker_sayısı + 1  # Artık değişken global olduğu için erişip değiştirilebilir.

seker_ekle()
print(seker_sayısı)


```

***Global Değişkenlerin Tehlikesi:*** Parametre kullanmak yerine dışarıdaki global değişkenleri fonksiyon içinde değiştirmek, kod büyüdükçe hatanın kaynağını bulmayı imkansızlaştırır. Aynı değişkene birden fazla fonksiyonun müdahale etmesi test yazmayı zorlaştırır ve eş zamanlı çalışan sistemlerde (multithreading) ciddi çakışmalara (race conditions) neden olur.

```python
def bahsis_ekle(hesap_tutari):
  yeni_tutar = hesap_tutari +300
  return yeni_tutar
print(bahsis_ekle(2000))
```
