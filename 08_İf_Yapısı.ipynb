```python
from os import ptsname
# Tek bir if-else bloğuyla---Birdem fazla koşul yazılabilir.
# Birden fazla şart varsa birden fazla koşul varsa bir if-else boğuyla
# birden fazla şart getirerek sorunu çözebiliriz.

yas = 20
kimlik_var = False

if yas >= 18 and kimlik_var:
  print("İşlam yapılabilir")
else:
  print("İşlem yapılamaz ")

# Tek şart yetmiyor ve şartları artırmamız gerekiyorsa -
# bu sistem gerçek hayata daha yakınlaşır.
# Güvenlik,Yetkilendirme ve doğrulam gibi süreçler bu şeklide şartlar artar.
```
