Şimdi, bu öğrendiklerimizi birleştiren yeni bir örnek yapalım: **Metin ve sayıları birleştirme** ve **kullanıcıdan alınan input'u dönüştürme**.

```python
# Kullanıcıdan isim ve yaş alalım
isim = input("Adınız nedir? ")
yas_str = input("Yaşınız kaç? ") # input her zaman string döndürür

# Yaşı string'den tam sayıya dönüştürelim
yas = int(yas_str)

# Kullanıcıya özel bir mesaj yazdıralım
print("Merhaba, " + isim + ". " + str(yas) + " yaşındasınız.")

# Yaşın 5 yıl sonraki halini hesaplayıp yazdıralım
gelecek_yas = yas + 5
print(f"5 yıl sonra {gelecek_yas} yaşında olacaksınız.")
```

```python
ogrenci_sayisi = int(input("öğrenci bilet sayısını giriniz. :"))
tam_sayısı = int(input("Tam bilet sayısını giriniz. :"))

ogrnci_fiyat = 75
tam_fiyat = 120

Toplam_bakiye = ogrenci_sayisi * ogrenci_sayisi + tam_fiyat * tam_sayısı

print("Toplam kasadaki para : {:.2f}".format(Toplam_bakiye))
```

```python
Gidilecek_yol = float(input("Gidilecek yolu giriniz (km): "))
Ortalama_yakit = float(input("Aracın 100 km'de yaktığı ortalama litreyi giriniz: "))
Benzin_fiyati = float(input("Güncel benzin fiyatını giriniz: "))

# Önce hesaplamalar
Harcanan_toplam_litre = (Gidilecek_yol / 100) * Ortalama_yakit
Toplam_Maliyet = Harcanan_toplam_litre * Benzin_fiyati

# Sonra çıktılar
print("Harcanan Yakıt: {:.1f} Litre".format(Harcanan_toplam_litre))
print("Toplam Tutar: {:.2f} TL".format(Toplam_Maliyet))
```

```python
Toplam_yemek_hesabi = float(input("Yemek için toplam ne kadar ücret ödendi?: "))
Masadaki_Musteri_Sayisi = int(input("Kaç kişi yemek yedi?: "))
Bahsis_Yuzdesi = float(input("Yemek için yüzde kaç bahşiş verilecek?: "))

Bahsis_Tutari = Toplam_yemek_hesabi * (Bahsis_Yuzdesi / 100)
Genel_Toplam = Toplam_yemek_hesabi + Bahsis_Tutari
Kisi_basi_dusen_tutar = Genel_Toplam / Masadaki_Musteri_Sayisi

# f-string kullanımı:
print(f"Kişi başı verilmesi gereken tutar: {Kisi_basi_dusen_tutar:.2f} TL")
```
