```python
"""
PROJE / MODÜL: Tesis & Şebeke Tüketim Analitik Motoru (Sprint 1)
YAZAR: İbrahim Halil Aslan | Elektrik-Elektronik Mühendisi & Veri Analisti
AÇIKLAMA:
    Bu script, sahadan alınan 3 aylık periyodik tüketim verilerini (kVA/kWh)
    kullanıcı girdisi olarak alır, veri tipini doğrular (float conversion),
    aritmetik ortalama hesabı yapar ve sonucu finansal/teknik standartlara
    uygun şekilde virgülden sonra 2 basamak hassasiyetle (f-string formatting)
    ekrana raporlar.

METRİKLER & YETKİNLİKLER:
    - Veri Doğrulama (Input Validation / Type Casting: float)
    - Aritmetik Mantık ve Operatör Önceliği (Parentheses & Division)
    - Metin Biçimlendirme & Raporlama (f-string Precision Formatting: :.2f)
"""

birinci_ay = float(input("ilk aydan olacak şekilde 1. ayın tüketim verilerini giriniz: "))
ikinci_ay = float(input("ilk aydan olacak şekilde 2. ayın tüketim verilerini giriniz: "))
ucuncu_ay = float(input("ilk aydan olacak şekilde 3. ayın tüketim verilerini giriniz: "))

uc_aylik_tuketim_ort = (birinci_ay + ikinci_ay + ucuncu_ay) / 3

print(f"İlk aydan itibaren 3 ayın tüketim ortalaması: {uc_aylik_tuketim_ort:.2f} kWh'dir")
```

```python
"""
PROJE / MODÜL: Üç Fazlı Şebeke & Trafo Aktif Güç (kW) Hesaplama Motoru (Sprint 1)
YAZAR: İbrahim Halil Aslan | Elektrik-Elektronik Mühendisi & Veri Analisti
AÇIKLAMA:
    Bu betik, sahadan alınan voltaj (V), akım (A) ve faz açısı (derece) parametrelerini
    kullanıcı girdisi olarak alır; trigonometrik dönüşümleri gerçekleştirip üç fazlı
    aktif güç formülünü uygulayarak sonucu kW cinsinden hesaplar ve raporlar.

MÜHENDİSLİK & KODLAMA STANDARTLARI (YETKİNLİKLER):
    1. Trigonometrik Dönüşüm (Trigonometry & Radians Conversion):
       - Python 'math' kütüphanesi trigonometrik hesaplamalarda radyan kabul ettiği için
         kullanıcıdan alınan faz açısı derece (°) cinsinden math.radians() fonksiyonu ile
         radyana çevrilmiş ve cos(phi) katsayısı çarpan haline getirilmiştir.
    2. Fiziksel Formülasyon & Matematiksel İşlem:
       - Üç fazlı aktif güç hesaplama formülü: P = (sqrt(3) * V * I * cos(phi)) / 1000 [kW]
       - Karekök işlemi için 'math.sqrt(3)' kütüphanesi entegre edilmiştir.
    3. Veri Biçimlendirme & Raporlama (Data Formatting):
       - 'f-string' ve ':.2f' hassasiyet belirteci kullanılarak çıktı finansal ve teknik
         standartlara uygun şekilde virgülden sonra 2 basamağa yuvarlanmıştır.
"""

import math

# 1. Kullanıcıdan Saha Ölçüm Değerlerinin Alınması (Type Casting: float)
voltaj = float(input("Voltaj değerini giriniz (V): "))
akim = float(input("Akım değerini giriniz (A): "))

# 2. Faz Açısının Alınması ve Radyan/Kosinüs Dönüşümü
aci_derece = float(input("Faz açısını derece olarak giriniz (Örn: 30): "))
cos_phi = math.cos(math.radians(aci_derece))

# 3. Üç Fazlı Aktif Güç Hesabı (kW Cinsinden)
aktif_guc = (math.sqrt(3) * voltaj * akim * cos_phi) / 1000

# 4. Raporlama (f-string Formatting)
print(f"Aktif gücünüz: {aktif_guc:.2f} kW")
```
