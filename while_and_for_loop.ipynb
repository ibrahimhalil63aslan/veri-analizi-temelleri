```python
products = ["Laptop", "Telefon", "Kulaklık"]
prices = [25000, 15000, 750]

# Başlangıçta en yüksek fiyatı 0, en pahalı ürünü boş kabul ediyoruz
en_yuksek_fiyat = 0
en_pahali_urun = ""

for urun, fiyat in zip(products, prices):
    # print(f"{urun}: {fiyat}")  # 1. ve 2. Adım: Ekrana yazdırıyoruz

    # 3. Adım: Kontrol ediyoruz
    if fiyat > en_yuksek_fiyat:
        en_yuksek_fiyat = fiyat      # Yeni en yüksek fiyatı hafızaya al
        en_pahali_urun = urun        # Yeni en pahalı ürünü hafızaya al

# Döngü bittikten sonra en son kalan kazananı yazdırıyoruz
print(f"En pahalı: {en_pahali_urun} - {en_yuksek_fiyat}")
```

```python
log_satirlari = [
    "INFO: Kullanıcı sisteme başarılı giriş yaptı.",
    "DEBUG: Geçici önbellek (cache) temizlendi.",
    "INFO: Veri tabanı sorgusu 0.2 saniyede tamamlandı.",
    "ERROR: Veritabanı bağlantı zaman aşımı (Timeout)! Sistem yanıt vermiyor.",  # Kritik hata
    "INFO: Oturum sonlandırıldı."
]

for log in log_satirlari:
    # 1. Aşama: DEBUG seviyesindeki sıradan logları atla (continue)
    if "DEBUG" in log:
        continue

    # 2. Aşama: Kritik bir hata (ERROR) bulduğumuzda döngüyü anında durdur (break)
    if "ERROR" in log:
        print(f"🚨 Kritik Hata Yakalandı: {log}")
        print("İşlem durduruldu, Sistem Yöneticisine alarm gönderiliyor...")
        break

    print(f"İncelenen Log: {log}")
```

```python
log_satirlari = [
    "INFO: Kullanıcı sisteme başarılı giriş yaptı.",
    "DEBUG: Geçici önbellek (cache) temizlendi.",
    "INFO: Veri tabanı sorgusu 0.2 saniyede tamamlandı.",
    "ERROR: Veritabanı bağlantı zaman aşımı (Timeout)! Sistem yanıt vermiyor.",  # 1. Kritik Hata
    "INFO: Kullanıcı sayfayı yeniledi.",
    "DEBUG: Session ID kontrol edildi.",
    "ERROR: Yetkilendirme hatası (401 Unauthorized)! İşlem reddedildi."          # 2. Kritik Hata
]

for log in log_satirlari:
    # 1. Aşama: DEBUG loglarını tamamen atla (continue)
    if "DEBUG" in log:
        continue

    # 2. Aşama: ERROR gördüğümüzde BREAK YAPMIYORUZ!
    # Hatayı raporlayıp döngünün sıradaki satıra geçmesini sağlıyoruz.
    if "ERROR" in log:
        print(f"🚨 Kritik Hata Yakalandı: {log}")
        continue  # Hatayı bastıktan sonra döngünün geri kalanını atlayıp bir sonraki loga geç

    # 3. Aşama: Kalan normal (INFO) logları ekrana yazdır
    print(f"✅ Normal Log: {log}")
```

```python
satislar = [150, 200, -50, 320, 0, 450, -10]
toplam_kazanc = 0

for satis in satislar:
    if satis <= 0:
        continue  # Negatif veya 0 olanları atla

    toplam_kazanc += satis  # Sadece geçerli satışları topla

print(f"Toplam Geçerli Satış Kazancı: {toplam_kazanc} TL")
```

```python
# Vardiya listesi
vardiyalar = ["Vardiya-1", "Vardiya-2", "Vardiya-3"]
vardiya_index = 0

# 1. Ana Kontrol: While döngüsü ile vardiyaları dönüyoruz
while vardiya_index < len(vardiyalar):
    aktif_vardiya = vardiyalar[vardiya_index]
    print(f"\n--- {aktif_vardiya} Başladı ---")

    # O vardiyaya ait dakikalık loglar
    loglar = ["ROUTINE_CHECK", "NORMAL", "CRITICAL_ERROR", "NORMAL"]

    # 2. Alt Kontrol: For döngüsü ile vardiya içindeki logları tarıyoruz
    for log in loglar:
        # A) continue kullanımı: Rutin kontrolleri atla
        if log == "ROUTINE_CHECK":
            continue

        # B) break kullanımı (İç döngüde): Kritik hata bulunca o vardiyayı kır/durdur
        if log == "CRITICAL_ERROR":
            print(f"🚨 {aktif_vardiya}'de Kritik Hata Tespit Edildi! Vardiya durduruluyor.")
            break

        print(f"✅ İşlem Logu: {log}")

    vardiya_index += 1
```

```python
# Şirketin vardiya paketleri
vardiyalar = ["Vardiya 1 Logları", "Vardiya 2 Logları"]
vardiya_index = 0

# 1. Ana Döngü: Vardiyaları dönmek için while kullanıyoruz
while vardiya_index < len(vardiyalar):
    aktif_vardiya = vardiyalar[vardiya_index]
    print(f"\n--- {aktif_vardiya} inceleniyor ---")

    # O vardiyaya ait log satırları
    log_satirlari = [
        "INFO: Sistem normal çalışıyor",
        "DEBUG: Geçici dosya silindi",
        "INFO: Kullanıcı girişi başarılı",
        "CRITICAL_SYSTEM_FAILURE: Bellek aşımı!", # Dikkat: Burada break patlamalı!
        "INFO: Bu satıra asla ulaşılmamalı"
    ]

    # 2. Alt Döngü: Log satırlarını taramak için for kullanıyoruz (Girintiye dikkat!)
    for log in log_satirlari:
        # A) DEBUG ise atla (continue)
        if "DEBUG" in log:
            continue

        # B) Kritik hata ise vardiyayı durdur (break)
        if "CRITICAL_SYSTEM_FAILURE" in log:
            print(f"🚨 {aktif_vardiya}'de Kritik Hata Tespit Edildi! Vardiya durduruluyor.")
            break

        print(f"İncelenen Log: {log}")

    # while döngüsünün sonsuza girmemesi için index'i artırıyoruz
    vardiya_index += 1
```

```python
toplam = 0

while True:
    # Kullanıcıdan sayı alıp integer'a çeviriyoruz
    sayi = int(input("Sayı giriniz: "))

    # 1. Kural: 0 girilirse döngüyü kır (break)
    if sayi == 0:
        break

    # 2. Kural: Negatifse toplama ekleme, adımı atla (continue)
    if sayi < 0:
        continue

    # 3. Kural: Pozitif sayıları toplam değişkeninde biriktir
    toplam += sayi

print(f"Girdiğiniz sayıların toplamı: {toplam}")
```
