```python
islem_tutarlari = [120,75,200,33,48,91,17,84]

cift_islemler =[]  # FİLTRELEME MANTIĞI

kontrol_sayisi=0  # Sayac

cift_islem_toplamı=0

MAX_KONTROL_LİMİTİ = 7 # Döngü sayısını verir.

for tutar in islem_tutarlari:
  kontrol_sayisi +=1
  if kontrol_sayisi > MAX_KONTROL_LİMİTİ:
    print("Kontrol limiti aşıldı. Analiz durdu")
    break
  if tutar % 2 == 0:
    cift_islemler.append(tutar)
    cift_islem_toplamı += tutar
  else:
    continue

print("Kontrol edilen işlem sayısı:",kontrol_sayisi)
print("Çift tutarlı işlemler :", cift_islemler)
print("Çift işlem toplamı:", cift_islem_toplamı)

```
