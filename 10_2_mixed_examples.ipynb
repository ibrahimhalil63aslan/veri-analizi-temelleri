```python
def gecerli_notlari_sec(liste):
    gecerli_olanlar = []

    for notu in liste:
        if 1 <= notu <= 10:
            gecerli_olanlar.append(notu)

    return gecerli_olanlar


gelen_notlar = [5, 12, 8, 0, 10, -3]
temiz_liste = gecerli_notlari_sec(gelen_notlar)

print(temiz_liste)
```

```python
def gecenlerin_filtresi(gecenler):
  sinav_nu = []

  for nu in gecenler:
    if 50 <= nu:
      sinav_nu.append(nu)
  return sinav_nu

sinav_notlari = [45, 60, 20, 85, 50, 10]
net_geçenler = gecenlerin_filtresi(sinav_notlari)

print(net_geçenler)
```

```python

benim_kutuphanem = [
    {"kitap_adi": "Dune", "durum": "okundu", "puan": 8},
    {"kitap_adi": "Sefiller", "durum": "okunacak", "puan": None},
    {"kitap_adi": "Kozmos", "durum": "okundu", "puan": 10}
]


def okunanlari_getir(liste):
    okunanlar_sepeti = []
    for kitap in liste:
        if kitap["durum"] == "okundu":
            okunanlar_sepeti.append(kitap)
    return okunanlar_sepeti

def isime_gore_ara(liste, aranan_kelime):
    bulunanlar = []
    for kitap in liste:
        if aranan_kelime.lower() == kitap["kitap_adi"].lower():
            bulunanlar.append(kitap)
    return bulunanlar

def ortalama_puan_hesapla(liste):
    puanlar = []
    for kitap in liste:
        if kitap["durum"] == "okundu" and kitap["puan"] != None:
            puanlar.append(kitap["puan"])

    if len(puanlar) == 0:
        return 0.0

    return sum(puanlar) / len(puanlar)

while True:
    print("\n--- KÜTÜPHANE YÖNETİM SİSTEMİ ---")
    print("1. Okunan Kitapları Listele")
    print("2. Kitap İsmine Göre Ara")
    print("3. Ortalama Puanı Hesapla")
    print("4. Çıkış")

    secim = input("Yapmak istediğiniz işlemi seçin (1-4): ")

    if secim == "1":
        sonuc = okunanlari_getir(benim_kutuphanem)
        print("\n--- OKUNAN KİTAPLAR ---")
        for k in sonuc:
            print(f"- {k['kitap_adi']} (Puan: {k['puan']})")

    elif secim == "2":
        aranan = input("Aradığınız kitabın adı: ")
        sonuc = isime_gore_ara(benim_kutuphanem, aranan)

        if len(sonuc) == 0:
            print("\n[!] Aradığınız kitap bulunamadı.")
        else:
            print("\n--- ARAMA SONUCU ---")
            for k in sonuc:
                print(f"Kitap: {k['kitap_adi']} | Durum: {k['durum']} | Puan: {k['puan']}")

    elif secim == "3":
        ort = ortalama_puan_hesapla(benim_kutuphanem)
        print(f"\n[+] Okunan kitapların puan ortalaması: {ort}")

    elif secim == "4":
        print("\nSistemden çıkılıyor. İyi günler!")
        break # Sonsuz döngüyü sonlandırır

    else:
        print("\n[!] Geçersiz seçim! Lütfen 1, 2, 3 veya 4 yazın.")
```
